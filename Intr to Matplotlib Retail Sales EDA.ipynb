#Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

#Dat Ingestion

```python
Spark_df = spark.sql("select* from default.retail_sales_dataset")
df = Spark_df.toPandas()

df.head()
```

```python
revenue_by_product = df.groupby("Product Category")["Total Amount"].sum()
print("Revenue by product" )
print(revenue_by_product)
```

```python
df.info()
```

#First visualization

```python
import matplotlib.pyplot as plt
import pandas as pd
revenue_by_product.plot(kind="bar")

plt.title("Revenue by product")
plt.xlabel("Product caretory")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
```

#Time Based Analysis(Line chart)

```python
df["date"] = pd.to_datetime(df["Date"])
df.info()
```

```python
revenue_over_time = df.groupby("date")["Total Amount"].sum()
```

```python
revenue_over_time.plot(kind="line")

plt.title("Revenue over time")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.show
```
